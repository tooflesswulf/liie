```python
import npoly2d as n2
import numpy as np
from scipy.linalg import expm

import yaml
import sympy as sym
import matplotlib.pyplot as plt
```

```python
N = 100
# Arm configuration
_gamma = np.zeros((N, 3))
_gamma[:, 0] = np.linspace(0, 1, N)  # Straight arm along x-axis
gamma = np.array([n2.rho(gi) for gi in _gamma])

sfx = ''  # [1,1,0] -> '', [1,1,1] -> '1'
if sfx == '':
    gab = np.diag([1, 1, 0])
elif sfx == '1':
    gab = np.diag([1, 1, 1])

with open(f'_orthopoly{sfx}.yaml', 'r') as f:
    polys_string = yaml.safe_load(f)
pp = [sym.sympify(ps) for ps in polys_string]
print('loaded bases')

def eval_poly(x):
    return np.array([pi.subs({sx: x}) for pi in pp]).astype(float)

sx = sym.symbols('t')
tt = np.linspace(0, 1, 100)
npoly = np.array([eval_poly(ti) for ti in tt])
```

```python
xi1 = np.zeros((N, 3))
xi1[:,2] = npoly[:,0]

xi2 = np.zeros((N, 3))
xi2[:,2] = npoly[:,1]

def metric(c1, c2):
    gamma = n2.shape_exp(c1*xi1 + c2*xi2)
    g00 = n2.inner_product(gamma, gab, xi1, xi1)
    g01 = n2.inner_product(gamma, gab, xi1, xi2)
    g11 = n2.inner_product(gamma, gab, xi2, xi2)
    print(g01, n2.inner_product(gamma, gab, xi2, xi1))
    return np.array([[g00, g01], [g01, g11]])
```
